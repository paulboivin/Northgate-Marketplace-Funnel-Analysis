```python
# Generate sessions table
# Target: ~75,000 sessions across Jan-Dec 2024
# Monthly distribution reflects realistic e-commerce seasonality

# Monthly session volume weights
monthly_weights = {
    1: 0.07,   # January — post-holiday slowdown
    2: 0.07,   # February — stable baseline
    3: 0.08,   # March — slight spring uplift
    4: 0.08,   # April — stable
    5: 0.08,   # May — stable
    6: 0.08,   # June — stable
    7: 0.08,   # July — summer browsing
    8: 0.08,   # August — summer browsing
    9: 0.08,   # September — back to school
    10: 0.08,  # October — pre-holiday buildup
    11: 0.11,  # November — peak season
    12: 0.11   # December — peak season
}

# Device distribution
device_types = ['Mobile', 'Desktop', 'Tablet']
device_weights = [0.65, 0.28, 0.07]

# Generate sessions
total_sessions = 75000
sessions_list = []
session_id = 1

for month, weight in monthly_weights.items():
    month_sessions = int(total_sessions * weight)
    
    for _ in range(month_sessions):
        # Random day within the month
        if month in [4, 6, 9, 11]:
            max_day = 30
        elif month == 2:
            max_day = 29  # 2024 is a leap year
        else:
            max_day = 31
            
        day = random.randint(1, max_day)
        session_date = datetime(2024, month, day).date()
        device = random.choices(device_types, 
                               weights=device_weights)[0]
        user_id = random.randint(1, 50000)
        
        sessions_list.append({
            'session_id': session_id,
            'session_date': session_date,
            'device_type': device,
            'user_id': user_id
        })
        session_id += 1

sessions = pd.DataFrame(sessions_list)

print(f"Sessions table shape: {sessions.shape}")
print(f"\nDate range: {sessions['session_date'].min()} "
      f"to {sessions['session_date'].max()}")
print(f"\nDevice distribution:")
print(sessions['device_type'].value_counts())
print(f"\nDevice distribution (%):")
print((sessions['device_type'].value_counts() / 
       len(sessions) * 100).round(1))
print(f"\nMonthly session distribution:")
sessions['month'] = pd.to_datetime(
    sessions['session_date']).dt.month
print(sessions.groupby('month').size())
print(f"\nSample rows:")
print(sessions.head(5).to_string())
```

```python
# Generate sessions table with natural monthly variation
# Monthly session volume targets with realistic variation

monthly_base = {
    1: 5250,    # January — post-holiday slowdown
    2: 5250,    # February — stable baseline  
    3: 6000,    # March — slight spring uplift
    4: 6000,    # April — stable
    5: 6000,    # May — stable
    6: 6000,    # June — stable
    7: 6000,    # July — summer browsing
    8: 6000,    # August — summer browsing
    9: 6000,    # September — back to school
    10: 6000,   # October — pre-holiday buildup
    11: 8250,   # November — peak season
    12: 8250    # December — peak season
}

# Add realistic variation of +/- 8% to each month
# This simulates natural fluctuation while preserving
# the overall seasonal pattern
monthly_sessions = {}
for month, base in monthly_base.items():
    variation = random.uniform(-0.08, 0.08)
    monthly_sessions[month] = int(base * (1 + variation))

print("Monthly session targets with variation:")
for month, count in monthly_sessions.items():
    print(f"  Month {month:02d}: {count:,} sessions")
print(f"  Total: {sum(monthly_sessions.values()):,} sessions")

# Device distribution
device_types = ['Mobile', 'Desktop', 'Tablet']
device_weights = [0.65, 0.28, 0.07]

# Generate sessions
sessions_list = []
session_id = 1

for month, count in monthly_sessions.items():
    for _ in range(count):
        if month in [4, 6, 9, 11]:
            max_day = 30
        elif month == 2:
            max_day = 29
        else:
            max_day = 31

        day = random.randint(1, max_day)
        session_date = datetime(2024, month, day).date()
        device = random.choices(device_types,
                               weights=device_weights)[0]
        user_id = random.randint(1, 50000)

        sessions_list.append({
            'session_id': session_id,
            'session_date': session_date,
            'device_type': device,
            'user_id': user_id
        })
        session_id += 1

sessions = pd.DataFrame(sessions_list)

# Add month column for analysis
sessions['month'] = pd.to_datetime(
    sessions['session_date']).dt.month

print(f"\nSessions table shape: {sessions.shape}")
print(f"Date range: {sessions['session_date'].min()} "
      f"to {sessions['session_date'].max()}")
print(f"\nDevice distribution (%):")
print((sessions['device_type'].value_counts() /
       len(sessions) * 100).round(1))
print(f"\nFinal monthly distribution:")
print(sessions.groupby('month').size())
```

```python
# Generate events table
# Each session progresses through funnel stages probabilistically
# Conversion rates vary by device type for realism

# Conversion rates by device type
# P(product_view | session) 
# P(add_to_cart | product_view)
# P(checkout | add_to_cart)
# P(purchase | checkout)

conversion_rates = {
    'Mobile': {
        'product_view': 0.62,
        'add_to_cart': 0.08,
        'checkout': 0.42,
        'purchase': 0.62
    },
    'Desktop': {
        'product_view': 0.70,
        'add_to_cart': 0.12,
        'checkout': 0.50,
        'purchase': 0.70
    },
    'Tablet': {
        'product_view': 0.66,
        'add_to_cart': 0.10,
        'checkout': 0.46,
        'purchase': 0.66
    }
}

# Category conversion modifiers
# Some categories convert better than others
category_modifiers = {
    'Books': 1.15,           # Easy purchase decision
    'Beauty': 1.05,          # Repeat purchases
    'Sports and Outdoors': 1.00,  # Baseline
    'Apparel': 0.95,         # Size uncertainty causes hesitation
    'Home and Kitchen': 0.90, # Higher consideration period
    'Electronics': 0.85      # Most research-intensive
}

events_list = []
event_id = 1

for _, session in sessions.iterrows():
    sid = session['session_id']
    device = session['device_type']
    session_date = session['session_date']
    rates = conversion_rates[device]

    # Base timestamp for this session
    base_hour = random.randint(6, 23)
    base_minute = random.randint(0, 59)
    base_ts = datetime.combine(
        session_date,
        datetime.min.time().replace(
            hour=base_hour, minute=base_minute))

    # Event 1 — session_start (every session)
    events_list.append({
        'event_id': event_id,
        'session_id': sid,
        'product_id': None,
        'event_type': 'session_start',
        'event_timestamp': base_ts,
        'price_at_event': None
    })
    event_id += 1

    # Event 2 — product_view (probabilistic)
    if random.random() < rates['product_view']:
        # Select a random product
        product = products.sample(1).iloc[0]
        pid = product['product_id']
        category = product['category']
        modifier = category_modifiers[category]

        # Price varies +/- 10% from base for realism
        price = round(product['base_price'] *
                     random.uniform(0.90, 1.10), 2)

        view_ts = base_ts + timedelta(
            seconds=random.randint(10, 120))

        events_list.append({
            'event_id': event_id,
            'session_id': sid,
            'product_id': int(pid),
            'event_type': 'product_view',
            'event_timestamp': view_ts,
            'price_at_event': price
        })
        event_id += 1

        # Event 3 — add_to_cart (probabilistic)
        if random.random() < rates['add_to_cart'] * modifier:
            cart_ts = view_ts + timedelta(
                seconds=random.randint(15, 180))

            events_list.append({
                'event_id': event_id,
                'session_id': sid,
                'product_id': int(pid),
                'event_type': 'add_to_cart',
                'event_timestamp': cart_ts,
                'price_at_event': price
            })
            event_id += 1

            # Event 4 — checkout (probabilistic)
            if random.random() < rates['checkout'] * modifier:
                checkout_ts = cart_ts + timedelta(
                    seconds=random.randint(30, 300))

                events_list.append({
                    'event_id': event_id,
                    'session_id': sid,
                    'product_id': int(pid),
                    'event_type': 'checkout',
                    'event_timestamp': checkout_ts,
                    'price_at_event': price
                })
                event_id += 1

                # Event 5 — purchase (probabilistic)
                if random.random() < rates['purchase'] * modifier:
                    purchase_ts = checkout_ts + timedelta(
                        seconds=random.randint(30, 180))

                    events_list.append({
                        'event_id': event_id,
                        'session_id': sid,
                        'product_id': int(pid),
                        'event_type': 'purchase',
                        'event_timestamp': purchase_ts,
                        'price_at_event': price
                    })
                    event_id += 1

events = pd.DataFrame(events_list)

# Convert product_id to nullable integer
events['product_id'] = pd.array(
    events['product_id'], dtype=pd.Int64Dtype())

print(f"Events table shape: {events.shape}")
print(f"\nEvent type distribution:")
print(events['event_type'].value_counts())
print(f"\nOverall funnel summary:")
stage_counts = events['event_type'].value_counts()
total_sessions = stage_counts['session_start']
for stage in ['session_start', 'product_view',
              'add_to_cart', 'checkout', 'purchase']:
    count = stage_counts[stage]
    pct = count / total_sessions * 100
    print(f"  {stage:<20} {count:>7,} ({pct:.1f}%)")
print(f"\nOverall conversion rate (session to purchase):")
print(f"  {stage_counts['purchase'] / total_sessions * 100:.2f}%")
```

```python
# Generate orders table
# Every purchase event generates exactly one order record

# Extract all purchase events
purchase_events = events[
    events['event_type'] == 'purchase'].copy()

# Generate orders
orders = pd.DataFrame({
    'order_id': range(1, len(purchase_events) + 1),
    'session_id': purchase_events['session_id'].values,
    'product_id': purchase_events['product_id'].values,
    'order_date': pd.to_datetime(
        purchase_events['event_timestamp']).dt.date.values,
    'quantity': [random.choices(
        [1, 2, 3, 4],
        weights=[0.70, 0.20, 0.07, 0.03])[0]
        for _ in range(len(purchase_events))],
    'unit_price': purchase_events['price_at_event'].values
})

# Calculate order value
orders['order_value'] = (
    orders['unit_price'] * orders['quantity']).round(2)

print(f"Orders table shape: {orders.shape}")
print(f"\nOrder quantity distribution:")
print(orders['quantity'].value_counts().sort_index())
print(f"\nOrder value statistics:")
print(orders['order_value'].describe().round(2))
print(f"\nTotal revenue generated:")
print(f"  ${orders['order_value'].sum():,.2f}")
print(f"\nAverage order value:")
print(f"  ${orders['order_value'].mean():.2f}")
print(f"\nMonthly order distribution:")
orders['month'] = pd.to_datetime(
    orders['order_date']).dt.month
print(orders.groupby('month').size())
print(f"\nSample rows:")
print(orders.head(5).to_string())
```
